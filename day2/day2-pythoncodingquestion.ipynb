```python
# Question 5: Reverse a number
num = int(input("Enter your nummber"))
reverse = 0

while num >0:
    digit = num %10
    reverse = reverse*10+digit
    num = num//10

print("Reverse:",reverse)
```

```python
# 6: Check whether a number is a palindrome
num = int(input("Enter number"))

original = num
reverse = 0

while num>0:
    digit = num %10
    reverse = reverse*10+digit
    num = num //10

if original ==reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Question 10: Check Armstrong Number
num = int(input())

original = num 
digits = len(str(num))
total = 0

while num >0:
    digit  = num % 10
    total = digit ** digits
    num //=10

if total == original:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
# ✅ 12. Factorial Using Recursion
num = int(input("Enter your number"))

def factorial(n):
    if n ==0 or n==1:
        return 1
    return n* factorial(n-1)


print("Factorial =",factorial(num))
```

```python
# ✅ 13. Print Prime Numbers Between 1 and 100
for num in range(2,101):

    prime = True

    for i in range(2,int(num*0.5)+1):
        if num % i ==0:
            prime =False
            break

    if prime:
        print(num,end=" ")
    
```

```python
# ✅ 14. Find GCD (Greatest Common Divisor)
a = int(input())
b = int(input())

while b !=0:
    a, b=b,a%b

print("GCD:",a)
```

```python
# ✅ 15. Find LCM (Least Common Multiple)
a = int(input())
b = int(input())

x = a
y = b

while y !=0:
    x,y =y,x%y

gcd = x

lcm = (a*b)//gcd
print("LCM:",lcm)
```

```python
# ✅ 17. Count Vowels and Consonants
text = input("Enter your string: ").lower()

vowels = 0
consonants = 0

for ch in text:
    if ch.isalpha():
        if ch in "aeiou":
            vowels+=1
        else:
            consonants+=1
print("Vowels =",vowels)
print("Consonants =",consonants)
```

```python
# ✅ 33. Count Uppercase and Lowercase Letters

text = input("Enter your text: ")

upper = 0
lower = 0

for ch in text:
    if ch.isupper():
        upper+=1
    elif ch.islower():
        lower+=1

print("Uppercase:",upper)
print("Lowercase:",lower)
```
