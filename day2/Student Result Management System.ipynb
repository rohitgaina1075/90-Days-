```python
# ===============================
# Student Result Management System
# ===============================

def calculate_total(marks):
    return sum(marks)

def calculate_percentage(total,subjects):
    return total / subjects

def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >=80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

def result_status(marks):
    for mark in marks:
        if mark < 33:
            return "Fail"
        return "PASS"

def display_result(name,marks):

    total = calculate_total(marks)
    percentage= calculate_percentage(total,len(marks))
    grade = calculate_grade(percentage)
    result = result_status(marks)

    print("\n" + "=" * 40)
    print("   Student Result")
    print("=" * 40)

    print(f"Name        :,{name}")
    print(f"Englsih     :,{marks[0]}")
    print(f"Math        :,{marks[1]}")
    print(f"Science     :,{marks[2]}")
    print(f"Computer    :,{marks[3]}")
    print(f"Hindi       :,{marks[4]}")

    print("-"*40)

    print(f"Total        :{total}/500")
    print(f"Percentage   :{percentage:2f}%")
    print(f"Grade        :{grade}")
    print(f"Result       :{result}")

    print("="*40)

print("=" * 40)
print("Student result mamagement system")
print("=" * 40)

name = input("Enter student Name: ")

marks = []

subjects=[
    "English",
    "Math",
    "Science",
    "computer",
    "Hindi"
]

for subject in subjects:
    while True:
        mark = int(input(f"Enter{subject} marks (0-100):"))

        if 0 <= mark <=100:
            marks.append(mark)
            break
        print("Invalid Marks! Enter between 0 to 100.")
display_result(name,marks)
```
