```python
# 10. What is Recursion?

# Answer:

# Recursion is a technique where a function calls itself until a stopping condition (base case) is met.

# Example
```

```python
def factorial(n):
    if n ==1:
        return 1

    return n* factorial(n-1)

# Used in:

# Factorial
# Fibonacci
# Tree traversal
# Divide and conquer algorithms
```

```python
# 15. What are the Four Pillars of OOP?
# 1. Encapsulation

# Wrapping data and methods together into a class.

# 2. Abstraction

# Showing only essential details while hiding implementation.

# 3. Inheritance

# A child class acquires properties and methods of a parent class.

# Example
```

```python
class animal:
    def sound(self):
        print("Animal sound")

class Dog(animal):
    pass

d = Dog()
d.sound()
```

```python
# 4. Polymorphism

# The same method name can behave differently depending on the object.

# Example
```

```python
class Cat:
    def sound(self):
        print("Meow")

class Dog:
    def sound(self):
        print("Bark")

animals = [Cat(), Dog()]

for animal in animals:
    animal.sound()
```

```python
# 29. What are Decorators?

# A decorator is a function that adds or modifies the behavior of another function without changing its original code.
```

```python
def decorator(func):
    def wrapper():
        print("Before")
        func()
        print("After")
    return wrapper

@decorator
def hello():
    print("Hello")

hello()

# Uses:

# Logging
# Authentication
# Timing functions
# Access control
```

```python
# 30. What are Generators?

# A generator is a special type of function that uses yield to produce values one at a time, instead of returning them all at once.

# Example
```

```python
def numbers():
    for i in range(5):
        yield i

for num in numbers():
    print(num)
```

```python
# Generator vs Normal Function
# Generator	Normal Function
# Uses yield	Uses return
# Produces values one at a time	    Returns the entire result at once
# More memory efficient	           Uses more memory for large data
# Suitable for large datasets	Suitable for smaller datasets
```
