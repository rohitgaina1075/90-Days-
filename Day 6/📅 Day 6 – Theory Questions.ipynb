```python
# ✅ 1. What are *args in Python?
# Answer

# *args allows a function to accept any number of positional arguments.

# Instead of passing a fixed number of arguments, you can pass as many as needed.

# Internally, *args stores all arguments in a tuple.
```

```python
def add(*args):
    return sum(args)

print(add(10, 20))
print(add(10, 20, 30, 40))
```

```python
# Interview Tip
# *args = Variable number of positional arguments
# Stored as a tuple
```

```python
# ✅ 2. What are **kwargs?
# Answer

# **kwargs allows a function to accept any number of keyword arguments.

# Internally, they are stored in a dictionary.
```

```python
def student(**kwargs):
    print(kwargs)

student(name="Rohit", age=21)
```

```python
# ✅ 3. Difference between *args and **kwargs
# *args	                         **kwargs
# Positional arguments	         Keyword arguments
# Stored as tuple	                 Stored as dictionary
# Uses *	                         Uses **
```

```python
def demo(*args, **kwargs):
    print(args)
    print(kwargs)
```

```python
# ✅ 4. What is a Lambda Function?
# Answer

# A lambda function is a small anonymous function written in one line.

# It is mainly used for short operations.
```

```python
square = lambda x: x*x

print(square(5))
```

```python
# ✅ 5. Difference between def and lambda
# def	                          lambda
# Named function	              Anonymous function
# Multiple statements	          Single expression
# Reusable	                  Best for short tasks
```

```python
# ✅ 6. What is map()?
# Answer

# map() applies a function to every element of an iterable and returns a map object (often converted to a list).
```

```python
numbers = [1, 2, 3]

result = list(map(lambda x: x*2, numbers))

print(result)
```

```python
# ✅ 7. What is filter()?
# Answer

# filter() returns only those elements that satisfy a given condition.
```

```python
numbers = [1,2,3,4,5]

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)
```

```python
# ✅ 8. What is reduce()?
# Answer

# reduce() repeatedly applies a function to combine all elements into a single value.

# It is available in the functools module.
```

```python
from functools import reduce

numbers = [1,2,3,4]

result = reduce(lambda x, y: x+y, numbers)

print(result)
```

```python
# ✅ 9. Difference between map(), filter(), and reduce()
# Function	            Purpose	                              Output
# map()	                Transform each element	              Iterable of transformed values
# filter()	            Keep elements matching a condition	  Iterable of selected values
# reduce()	            Combine all elements	              Single value
```

```python
# ✅ 10. What is Recursion?
# Answer

# Recursion is a technique where a function calls itself to solve a smaller version of the same problem.

# Every recursive function must have a base case to stop the recursion.
```

```python
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```

```python
# ✅ 11. What is a Base Case?
# Answer

# A base case is the condition that stops recursion.

# Without a base case, recursion continues until Python raises a RecursionError.
```

```python
if n == 0:
    return 1

# This is the base case in the factorial example.
```

```python
# ✅ 12. What is an Iterator?
# Answer

# An iterator is an object that allows you to traverse elements one by one.

# It implements the methods:
# __iter__()
# __next__()
```

```python
numbers = [10, 20, 30]

it = iter(numbers)

print(next(it))
print(next(it))
```

```python
# ✅ 13. What is a Generator?
# Answer

# A generator is a special type of iterator created using the yield keyword.

# It generates values one at a time, making it memory efficient.
```

```python
def count():
    yield 1
    yield 2
    yield 3

for value in count():
    print(value)
```

```python
✅ 14. Difference between Iterator and Generator
Iterator	                                  Generator
Created with iter() or a custom class	      Created using yield
More code to implement	                       Easier to write
Must implement __iter__() and __next__()	   Python handles these automatically
Can be less convenient	                       More memory efficient
```
