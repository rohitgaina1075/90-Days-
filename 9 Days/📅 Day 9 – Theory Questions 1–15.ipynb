```python
# ✅ 1. What is Exception Handling?

# Answer:
# Exception Handling Python में runtime errors को handle करने का तरीका है, ताकि program अचानक crash न हो।

# मुख्य keywords:

# try
# except
# else
# finally
# raise
```

```python
# ✅ 2. What is an Exception?

# Answer:
# Exception एक ऐसी runtime घटना है जो program के normal execution को interrupt करती है।

# Example:

# print(10 / 0)

# यह ZeroDivisionError देगा।
```

```python
# ✅ 3. What is the purpose of try?

# Answer:
# try block में वह code लिखा जाता है जिसमें exception आने की संभावना होती है।

# try:
#     result = 10 / 0
```

```python
# ✅ 4. What is the purpose of except?

# Answer:
# except block exception को catch और handle करता है।

# try:
#     result = 10 / 0

# except ZeroDivisionError:
#     print("Cannot divide by zero")
```

```python
# ✅ 5. What is else in Exception Handling?

# Answer:
# else block तब execute होता है जब try block successfully execute हो जाए और कोई exception न आए।

# try:
#     result = 10 / 2

# except ZeroDivisionError:
#     print("Error")

# else:
#     print(result)
```

```python
# ✅ 6. What is finally?

# Answer:
# finally block cleanup code के लिए use होता है और सामान्यतः exception आने या न आने की स्थिति में execute होता है।

# try:
#     print("Hello")

# finally:
#     print("Finished")
```

```python
# ✅ 7. Difference Between except and finally
# except	                                          finally
# Exception handle करता है	                          Cleanup करता है
# Specific exception के लिए run हो सकता है	          सामान्यतः हमेशा execute होता है
# Error handling	                                  Resource cleanup
```

```python
# ✅ 8. What is raise?

# Answer:
# raise keyword का इस्तेमाल manually exception generate करने के लिए होता है।

# age = 15

# if age < 18:
#     raise ValueError("Age must be 18 or above")
```

```python
# ✅ 9. What is a Custom Exception?

# Answer:
# Custom Exception एक user-defined exception है जिसे हम अपनी application की specific requirements के लिए बनाते हैं।

# class InsufficientBalanceError(Exception):
#     pass

# फिर:

# raise InsufficientBalanceError("Insufficient balance")
```

```python
# ✅ 10. What is assert?

# Answer:
# assert किसी condition को check करता है। अगर condition False होती है, तो AssertionError raise होता है।

# age = 20
# assert age >= 18

# यह successfully execute होगा।
# लेकिन:

# age = 15
# assert age >= 18

# तो AssertionError आएगा।
# Interview tip: assert को user-input validation या production business logic के लिए primary validation mechanism नहीं मानना चाहिए, क्योंकि Python optimized mode में assertions disable हो सकती हैं।
```

```python
# ✅ 11. What is Exception as e?

# Answer:
# as e exception object को variable e में store करता है।

# try:
#     int("abc")

# except ValueError as e:
#     print(e)

# यह actual error message print कर सकता है।
```

```python
# ✅ 12. What is the difference between Error and Exception?

# Answer:
# Python terminology में दोनों शब्द अक्सर loosely इस्तेमाल होते हैं, लेकिन technically Python में runtime problems generally exceptions के रूप में represented होती हैं।

# Example:

# ValueError
# TypeError
# IndexError
# KeyError

# ये exception classes हैं।
```

```python
# ✅ 13. What is Exception Hierarchy?

# Answer:
# Python exceptions inheritance hierarchy में organized होती हैं।

# Example:

# BaseException
#     │
#     └── Exception
#           │
#           ├── ValueError
#           ├── TypeError
#           ├── LookupError
#           │     ├── IndexError
#           │     └── KeyError
#           └── ...

# इसलिए specific exception को catch करना generally बेहतर होता है।
```

```python
# ✅ 14. What is the difference between raise and assert?
# raise	                                                   assert
# Manually exception raise करता है	                           Condition check करता है
# Business/application logic में useful	                       Development/debugging में useful
# Different exception types raise कर सकता है	               False होने पर AssertionError
# Example: raise ValueError(...)	                           Example: assert x > 0
```

```python
# ✅ 15. Why is Exception Handling important?

# Answer:
# Exception handling program को more reliable, robust और user-friendly बनाता है।

# Benefits:
# Program crashes को control करना
# Meaningful error messages देना
# Invalid input handle करना
# Resources safely close करना
# Debugging आसान करना
# Production applications को stable बनाना

# Example:
# try:
#     amount = float(input("Enter amount: "))

#     if amount <= 0:
#         raise ValueError("Amount must be positive")

# except ValueError as e:
#     print("Invalid input:", e)
```
