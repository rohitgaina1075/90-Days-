```python
# ✅ 9. Create a Class for a Car

class Car:

    def __init__(self,brand,model):
        self.brand= brand
        self.model = model

    def display(self):
        print("Brand:",self.brand)
        print("Model:",self.model)

car = Car("Toyota","BMW")

car.display()
```

```python
# ✅ 10. Create a Bank Account Class

class BankAccount:

    def __init__(self,name,balance):
        self.name = name
        self.balance = balance

    def deposit(self,amount):
        self.balance +=amount

    def withdraw(self,amount):
        if amount <=self.balance:
            self.balance -=amount
        else:
            print("Insufficient Balance")

    def show_balance(self):
        print("Balance=",self.balance)

account = BankAccount("Rohit",1000)

account.deposit(5000)

account.withdraw(3000)

account.show_balance()
```

```python
# ✅ 11. Class Variable
# What is a Class Variable?

# A class variable is shared by all objects of the class.
```

```python
class student:

    school = "ABC school"

    def __init__(self,name):
        self.name = name

student1 = student("Rohit")
student2 = student("Amit")

print(student1.school)
print(student2.school)
```

```python
# ✅ 12. Instance Variable vs Class Variable

class student:

    school = "ABC school"

    def __init__(self,name):
        self.name = name

s1 = student("Rohit")
s2 = student("Amit")

print(s1.name)
print(s2.name)

print(s1.school)
print(s2.school)

```

```python
# ✅ 14. Class Method
# What is a Class Method?

# A class method works with class variables.
```

```python
class student:

    school = "ABC School"

    @classmethod
    def change_school(cls,name):
        cls.school = name

student.change_school("XYZ School")

print(student.school)
```

```python
# ✅ 15. Static Method
# What is a Static Method?

# A static method does not use self or cls.
```

```python
class Math:

    @staticmethod
    def add(a,b):
        return a + b

print(Math.add(10,20))

# Interview Tip

# Use a static method when the function doesn't depend on object or class data.
```

```python
# ✅ 16. Single Inheritance

class Animal:

    def sound(self):
        print("Animal Sound")

class Dog(Animal):

    def bark(self):
        print("Dog Brak")

dog = Dog()

dog.sound()
dog.bark()
```

```python
# ✅ 17. Method Overriding

class Animal:

    def sound(self):
        print("Animal Sound")

class Dog(Animal):

    def sound(self):
        print("Dog braks")

dog = Dog()    
dog.sound()

# The child class replaces the parent's implementation.
```

```python
# ✅ 18. Use super()

class Animal:

    def __init__(self):
        print("Animal Constructor")

class Dog(Animal):

    def __init__(self):
        super().__init__()
        print("Dog Constructor")

dog = Dog()

# super() calls methods or constructors from the parent class.
```

```python
# ✅ 19. Multilevel Inheritance

class Grandfather:

    def house(self):
        print("Grandfather's House")

class Father(Grandfather):

    def car(self):
        print("Father's Car")

class Son(Father):

    def bike(self):
        print("Son's Bike")

s = Son()

s.house()
s.car()
s.bike()
```

```python
# ✅ 20. Multiple Inheritance
class Father:

    def money(self):
        print("Father's Money")

class Mother:

    def love(self):
        print("Mother's Love")

class Child(Father,Mother):
    pass

c = Child()

c.money()
c.love()
```

```python
# ✅ 21. Encapsulation (Private Variable)
# What is Encapsulation?

# Encapsulation means hiding data from direct access.

# In Python, prefixing an attribute with __ makes it name-mangled (commonly called private).
```

```python
class Student:

    def __init__(self):
        self.__marks = 90

    def show_marks(self):
        print(self.__marks)

student = Student()

student.show_marks()
```

```python
# ✅ 22. Getter Method
class Student:

    def __init__(self):
        self.__marks = 95

    def get_marks(self):
        return self.__marks

student = Student()

print(student.get_marks())
```

```python
# ✅ 23. Setter Method

class Student:

    def __init__(self):
        self.__marks = 0

    def set_marks(self,marks):
        self._marks = marks

    def get_marks(self):
        return self.__marks

student = Student()

student.set_marks(88)

print(student.get_marks())
```

```python
# ✅ 24. Polymorphism

# Different classes use the same method name with different behavior.
```

```python
class Dog:

    def sound(self):
        print("Bark")

class Cat:

    def sound(self):
        print("Meow")

animals = [Dog(),Cat()]

for animal in animals:
    animal.sound()
```

```python
# ✅ 25. Abstraction Using ABC'

from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):

    def start(self):
        print("Car Started")

car = Car()
car.start()
```

```python
# ✅ 27. Constructor Overriding with super()

class Animal:

    def __init__(self):
        print("Animal Construtor")

class Dog(Animal):

    def __init__(self):
        super().__init__()
        print("Dog Construtor")

dog = Dog()
```

```python
# ✅ 28. Operator Overloading

class Number:

    def __init__(self,value):
        self.value = value

    def __add__(self, other):
        return self.value +other.value

n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
```
