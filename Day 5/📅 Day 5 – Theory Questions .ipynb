```python
# ✅ 1. What is Object-Oriented Programming (OOP)?
# Answer

# Object-Oriented Programming (OOP) is a programming paradigm that organizes code into objects and classes.

# Instead of writing only functions, OOP groups data (attributes) and behavior (methods) together.

# Advantages
# Code Reusability
# Better Organization
# Easier Maintenance
# Improved Security
# Real-World Modeling
```

```python
class Student:
    def study(self):
        print("Student is studying")

student = Student()
student.study()
```

```python
# ✅ 2. What is a Class?
# Answer

# A class is a blueprint or template used to create objects.

# It defines:

# Variables (attributes)
# Functions (methods)
```

```python
class Car:
    pass
```

```python
# ✅ 3. What is an Object?
# Answer

# An object is an instance of a class.

# Objects have:

# State (attributes)
# Behavior (methods)
```

```python
class Car:
    pass

car1 = Car()
car2 = Car()
```

```python
# ✅ 4. Difference Between Class and Object
# Class	                                           Object
# Blueprint	                                     Instance of a class
# Logical entity	                               Physical entity in memory
# Created using class	                          Created by calling the class
# Doesn't occupy memory for object data	      Occupies memory
```

```python
class Student:
    pass

student = Student()
```

```python
# ✅ 5. What is a Constructor?
# Answer

# A constructor is a special method that runs automatically when an object is created.

# In Python, the constructor is:
```

```python
class Student:
    def __init__(self):
        print("Constructor Called")

student = Student()
```

```python
# ✅ 6. What is self?
# Answer

# self refers to the current object of the class.

# It is used to access:

# Instance variables
# Instance methods
```

```python
class Student:
    def __init__(self, name):
        self.name = name
```

```python
# When you create:

# student = Student("Rohit")

# Python internally calls:

# Student.__init__(student, "Rohit")

# So self becomes student.
```

```python
# ✅ 7. What is an Instance Variable?
# Answer

# An instance variable belongs to each object individually.

# Each object has its own copy.
```

```python
class Student:
    def __init__(self, name):
        self.name = name

# Different objects can have different values for name.
```

```python
# ✅ 8. What is a Class Variable?
# Answer

# A class variable belongs to the class and is shared by all objects.
```

```python
class Student:
    school = "ABC School"

# Every object accesses the same school value unless it is changed.
```

```python
# ✅ 9. Difference Between Instance Variable and Class Variable
# Instance Variable	                 Class Variable
# Belongs to an object	             Belongs to the class
# Defined using self                   Defined directly inside the class
# Separate copy for each object	     Shared by all objects
```

```python
# ✅ 10. What is an Instance Method?
# Answer

# An instance method works with object data and always has self as the first parameter.

# Example
```

```python
class Student:
    def greet(self):
        print("Hello!")

student = Student()
student.greet()
```

```python
# ✅ 11. What is a Class Method?
# Answer

# A class method works with class variables instead of instance variables.

# It uses the @classmethod decorator and cls parameter.
```

```python
class Student:
    school = "ABC"

    @classmethod
    def change_school(cls, name):
        cls.school = name
```

```python
# ✅ 12. What is a Static Method?
# Answer

# A static method does not use self or cls.

# It is used for utility functions related to the class.
```

```python
class Math:
    @staticmethod
    def add(a, b):
        return a + b

print(Math.add(5, 3))
```

```python
✅ 13. Difference Between Instance, Class, and Static Methods
Instance Method	|  Class Method	|  Static Method
Uses self	    | Uses cls	    | Uses neither
Accesses object data	|Accesses class data	Utility function
Called on an object	Called on the class or object	Called on the class or object
```
