```python
# ✅ 25. What is Recursion?
# ✅ Answer: A function calling itself
```

```python
def countdown(n):
    if n ==0:
        return 
    print(n)
    countdown(n-1)

countdown(5)
```

```python
# ✅ 26. Which keyword creates a generator?
# ✅ Answer: yield
```

```python
def num():
    yield 1
    yield 2

for n in num():
    print(n)
```
