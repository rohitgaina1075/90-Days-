```python
# ✅ 18. Reverse a List Without Using reverse()

n = [1,2,3,4,5,5,6,7]

reverse_list = []

for i in range(len(n)-1,-1,-1):
    reverse_list.append(n[i])

# Loop starts from the last index and moves backward.

print(reverse_list)
```

```python
# ✅ 25. Count Character Frequency
text = input("Enter yur string:")

freq = {}

for ch in text:
    if ch in freq:
        freq[ch] +=1
    else:
        freq[ch]=1

print(freq)
```

```python
# ✅ 26. Function to Check Prime Number
def is_prime(num):

    if num <2:
        return False

    for i in range(2,num):
        if num %2 ==0:
            return False

    return True

n = int(input("Enter your number"))
if is_prime(n):
    print("Prime number")
else:
    print("Not prime number")
```

```python
# ✅ 27. Function to Find Factorial

def factorial(n):

    fact = 1

    for i in range(1,n+1):
        fact*=i

    return fact

num = int(input("Enter number:"))

print(factorial(num))
```

```python
# ✅ 28. Function to Find Maximum of Three Numbers
def maximum(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=c and b>=a:
        return b
    else:
        return c

print(maximum(23,4555,6))
```

```python
# ✅ 29. Function to Reverse a String
def reverse(text):

    result = ""

    for ch in text:
        result = ch +result

    return result

print(reverse("pyhton is good"))
```

```python
# ✅ 30. Function to Count Vowels

def count_vowels(text):
    count = 0

    for ch in text.lower():

        if ch in "aeiou":
            count+=1

    return count
print(count_vowels("hello"))
```

```python
# ✅ 31. Inverted Star Pattern
rows = 5

for i in range(rows,0,-1):
    print("*" *i)
```

```python
# ✅ 32. Hollow Square Pattern
rows = 5

for i in range(rows):
    for j in range(rows):
        if i ==0 or i ==rows -1 or j ==0 or j ==rows-1:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
# ✅ 33. Floyd's Triangle
num = 1

for i in range(1,6):

    for j in range(i):
        print(num,end=" ")
        num+=1
    print()
```

```python
# ✅ 34. Number Pyramid

rows = 5

for i in range(1,rows+1):
    print(" "* (rows-1),end="")

    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
# ✅ 34. Number Pyramid

rows = 5

for i in range(1,rows+1):
    print(" "* (rows-i),end="")

    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
# ✅ 35. Diamond Star Pattern
rows = 5

for  i in  range(rows):

    print(" "*(rows - i - 1),end="")
    print("* "*(i+1))

```
