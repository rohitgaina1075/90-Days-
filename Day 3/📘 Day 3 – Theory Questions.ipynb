```python
# ✅ 1. What is a Linked List?

# A Linked List is a linear data structure where each element (called a node) stores:

# Data
# Address (reference) of the next node

# Unlike arrays, linked lists are not stored in contiguous memory.

# Structure
# +------+-------+     +------+-------+     +------+------+
# |  10  |   o-------> |  20  |   o-------> |  30  | None |
# +------+-------+     +------+-------+     +------+------+
```

```python
class listnode:
    def __init__(self,val=0,next=None):
        self.val = val
        self.nextt = next
```

```python
# Advantages
# Dynamic size
# Easy insertion and deletion
# Disadvantages
# No direct indexing
# Uses extra memory for pointers
```

```python
# ✅ 2. Difference Between Array and Linked List
# Array	                                     Linked List
# Contiguous memory	                         Non-contiguous memory
# Fixed size (traditional arrays)	         Dynamic size
# Fast indexing (O(1))	                     Sequential access (O(n))
# Slow insertion/deletion	                 Fast insertion/deletion
# Less memory	                             More memory (stores pointers)
```

```python
# ✅ 3. What is Linear Search?

# Linear Search checks each element one by one until it finds the target.
```

```python
n = [1,2,3,4,5,6,7]

target = 5

for i in range(len(n)):
    if n[i] == target:
        print(i)
```

```python
# ✅ 4. What is Binary Search?

# Binary Search repeatedly divides a sorted array into halves until the target is found.
```

```python
# ✅ 5. When Should Binary Search Be Used?

# Binary Search should be used only when the data is sorted.

# Conditions
# Array/List must be sorted.
# Random access should be available (works well with arrays).
# Example
# 10 20 30 40 50 60

# Works perfectly.
```

```python
# ✅ 6. Time Complexity of Linear Search
# Best Case
# O(1)

# Target is the first element.

# Worst Case
# O(n)

# Target is the last element or not present.

# Average Case
# O(n)
```

```python
# ✅ 7. Time Complexity of Binary Search
# Best Case
# O(1)

# Middle element is the target.

# Worst Case
# O(log n)

# The search space is halved every iteration.
```

```python
# ✅ 12. What is Recursion?

# Recursion is a technique where a function calls itself until a base condition is reached.
```

```python
def countdown(n):
    if n ==0:
        return
    print(n)
    countdown(n-1)

countdown(5)
```

```python
# Advantages
# Cleaner code for recursive problems.
# Useful for trees and graphs.
# Disadvantages
# More memory usage.
# Can cause stack overflow if the base case is missing.
```
