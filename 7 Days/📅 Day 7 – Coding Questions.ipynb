```python
# ✅ 11. Convert Dictionary to JSON
import json

student = {
    "name":"Rohit",
    "age":23,
    "Course":"Data Science"
}

json_data = json.dumps(student)

print(json_data)

# json.dumps() converts a Python object → JSON string.
# dumps means dump string.
```

```python
# ✅ 12. Convert JSON to Dictionary
import json

data = '{"Name":"Rohit","age":23}'

student = json.loads(data)

print(student)
print(type(student))

# json.loads() converts a JSON string → Python dictionary.
# loads means load string.
```

```python
# ✅ 18. Use defaultdict
from collections import defaultdict

student = defaultdict(int)

student["Math"]+=10

print(student)

# If the key doesn't exist, defaultdict(int) starts it at 0.
```

```python
# ✅ 19. Use deque
from collections import deque

queue = deque([1,2,3])

queue.append(4)

queue.appendleft(0)

print(queue)
```

```python
# ✅ 20. Remove Elements from deque
from collections import deque

queue = deque([1,2,3,4])

queue.pop()

queue.popleft()

print(queue)

# pop() removes from the right end.
# popleft() removes from the left end.
```
