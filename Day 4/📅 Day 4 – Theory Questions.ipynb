```python
# ✅ 7. What is Exception Handling?
# Answer

# Exception handling prevents a program from crashing when an error occurs.

# It allows us to handle errors gracefully.
```

```python
try:
    print(10 / 0)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
# ✅ 8. Explain try, except, else, and finally
# try

# Contains code that may produce an error.

# except

# Handles the error.

# else

# Runs only if no exception occurs.

# finally

# Always runs, whether an exception occurs or not.
```

```python
try:
    print(10 / 2)

except:
    print("Error")

else:
    print("No Error")

finally:
    print("Finished")
```
