```python
# ✅ 1. Check Whether a Number is an Armstrong Number
# What is an Armstrong Number?

# An Armstrong number is a number whose digits raised to the power of the number of digits equal the number itself.
```

```python
n = int(input("Enter your number: "))

temp = n
digits = len(str(n))
total = 0

while temp > 0:
    digit = temp % 10
    total +=digit ** digits
    temp //=10

if temp == n :
    print("Armstrong number")
else:
    print("Not Armstrong")
```

```python
# ✅ 2. Check Whether a Number is a Strong Number
# What is a Strong Number?

# A Strong number is a number where the sum of the factorials of its digits equals the number.
```

```python
n = int(input("Enter number:"))

temp = n
total = 0

while temp >0:

    digit = temp % 10

    fact = 1

    for i in range(1,digit+1):
        fact*=i

    total +=fact

    temp //= 10

if total ==n:
    print("Strong Number")
else:
    print("Not Strong number")
```

```python
# ✅ 3. Find the GCD of Two Numbers
# GCD

# Greatest Common Divisor
```

```python
a = int(input("Enter your first number:"))
b = int(input("Enter your second number:"))

small = min(a,b)

gcd = 1

for i in range(1, small +1):

    if a % i==0 and b%i ==0:
        gcd = i

print("GCD",gcd)
```

```python
# ✅ 4. Find the LCM of Two Numbers
a = int(input())
b = int(input())

small = min(a,b)

gcd = 1

for i in range(1,small+1):
    if a%i==0 and b %i==0:
        gcd = 1

lcm = (a*b) //gcd

print("LCM",lcm)
```

```python
# ✅ 26. Palindrome Using Function

def palindrome(text):
    if text  == text[::-1]:
        return True
    return False

word = input("Enter word: ")

if palindrome(word):
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
# ✅ 28. Largest Element Using Function
def largest(numbers):

    maximum = numbers[0]

    for num in numbers:

        if num > maximum:
            maximum = num

    return maximum


numbers = [10, 20, 90, 50,100]

print(largest(numbers))
```

```python
# ✅ 30. Leap Year Using Function
# Rule

# A year is a leap year if:

# Divisible by 400, or
# Divisible by 4 but not by 100.
```

```python
def leap(year):

    if year % 400 ==0:
        return True

    elif year % 4 == 0 and year % 100 !=0:
        return True

    return False

year = int(input("Leap Year:"))

if leap(year):
    print("Leap year")
else:
    print("Not Leap year")
```

```python
# ✅ 32. Left Triangle Pattern
rows = 5

for i in range(1,rows +1):
    print(" " * (rows - i)+ "*"* i)
```

```python
# ✅ 33. Hollow Rectangle Pattern

rows = 4
cols = 5

for i in range(rows):

    for j in range(cols):
        if i ==0 or i == rows - 1 or j == 0 or j == cols - 1:
            print("*",end="")
        else:
            print(" ",end="")
    print()
```
