```python
# ✅ 7. What is a Class Variable?
# Answer

# A class variable is shared by all objects of the class.
```

```python
class Student:

    school = "ABC School"

# All Student objects access the same school value unless it is overridden.
```

```python
# ✅ 8. Difference Between Instance Variable and Class Variable
# Instance Variable	                  Class Variable
# Belongs to one object	              Shared by all objects
# Defined using self	                  Defined directly in the class
# Different value for each object	       Same value for all objects (unless changed)
```

```python
# ✅ 9. What is Inheritance?
# Answer

# Inheritance allows one class (child class) to reuse the attributes and methods of another class (parent class).

# Example
```

```python
class Animal:

    def sound(self):
        print("Animal Sound")

class Dog(Animal):
    pass

# Dog automatically inherits the sound() method.

# Benefits:

# Code reuse
# Reduced duplication
# Easier maintenance
```

```python
# ✅ 10. Types of Inheritance in Python

# Python supports:

# Single Inheritance
# Multiple Inheritance
# Multilevel Inheritance
# Hierarchical Inheritance

# Note: Python does not directly support hybrid inheritance as a separate type, though hybrid structures can be created by combining supported inheritance patterns.
```

```python
# ✅ 11. What is Method Overriding?
# Answer

# Method overriding occurs when a child class provides its own implementation of a method that already exists in the parent class.
```

```python
class Animal:

    def sound(self):
        print("Animal Sound")

class Dog(Animal):

    def sound(self):
        print("Dog Bark")
```

```python
# ✅ 12. What is super()?
# Answer

# super() is used to call methods or constructors of the parent class.
```

```python
class Animal:

    def __init__(self):
        print("Animal")

class Dog(Animal):

    def __init__(self):
        super().__init__()
        print("Dog")
```

```python
# ✅ 13. What is Polymorphism?
# Answer

# Polymorphism means one interface, many implementations. The same method call can produce different behavior depending on the object.
```

```python
class Dog:

    def sound(self):
        print("Bark")

class Cat:

    def sound(self):
        print("Meow")

animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
```

```python
# ✅ 14. What is Encapsulation?
# Answer

# Encapsulation is the process of hiding data and providing controlled access to it.
```

```python
class Student:

    def __init__(self):
        self.__marks = 90

    def get_marks(self):
        return self.__marks

# Benefits:

# Data protection
# Better control over access
# Easier maintenance
```
