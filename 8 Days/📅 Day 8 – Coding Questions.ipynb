```python
# ✅ 1. Create a Class and Object
class Student:

    def display(self):
        print("Welcome to Python")

s = Student()
s.display()

# 📖 Explanation
# class Student: creates a class.
# s = Student() creates an object.
# s.display() calls the method.
```

```python
# ✅ 2. Create a Constructor (__init__)
class Student:

    def __init__(self):
        print("Constructor called")

s = Student()

# __init__() is a constructor.
# It runs automatically when an object is created.
```

```python
# ✅ 3. Constructor with Parameters
class student:

    def __init__(self,name,age):
        self.name = name
        self.age = age

    def show(self):
        print(self.name)
        print(self.age)

s= student("Rohit",23)
s.show()

# self.name and self.age are instance variables.
# Each object can store different values.
```

```python
# ✅ 4. Create Multiple Objects
class student:

    def __init__(self,name):
        self.name = name

    def show(self):
        print(self.name)

s1 = student("Rohit")
s2 = student("Amit")

s1.show()
s2.show()
```

```python
# ✅ 5. Class Variable
class Student:

    school = "ABC School"

    def __init__(self,name):
        self.name = name

    def show(self):
        print(self.name)
        print(Student.school)

s = Student("Rohit")
s.show()

# school is shared by all objects.
# Access it using Student.school.
```

```python
# ✅ 6. Instance Variable vs Class Variable
class Student:

    school = "ABC"

    def __init__(self,name):
        self.name = name

s1 = Student("Rohit")
s2 = Student("amit")

print(s1.name)
print(s2.name)

print(Student.school)
```

```python
# ✅ 7. Instance Method
class Student:

    def greet(self):
        print("Hello Student")

s = Student()

s.greet()
# greet() is an instance method.
# It works with an object.
```

```python
# ✅ 8. Modify Object Attributes
class Student:

    def __init__(self, name):
        self.name = name

s = Student("Rohit")

print(s.name)

s.name = "Rahul"

print(s.name)
```

```python
# ✅ 9. Delete Object Attribute
class Student:

    def __init__(self):
        self.name = "Rohit"

s = Student()

print(s.name)

del s.name

# fter del s.name, trying to access s.name will raise an AttributeError because the attribute no longer exists.
```

```python
# ✅ 10. Delete an Object
class Student:
    pass

s = Student()

del s

# del s removes the reference to the object.
# If no references remain, Python's garbage collector can free the memory.
```

```python
# ✅ 11. Single Inheritance
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    pass


d = Dog()
d.sound()

# Dog inherits from Animal.
# Dog automatically gets access to the sound() method.
```

```python
# ✅ 12. Add a New Method in Child Class
class Animal:

    def sound(self):
        print("Animal Sound")


class Dog(Animal):

    def bark(self):
        print("Dog Barks")


d = Dog()

d.sound()
d.bark()
```

```python
# ✅ 13. Method Overriding
class Animal:

    def sound(self):
        print("Animal Sound")


class Dog(Animal):

    def sound(self):
        print("Dog Barks")


d = Dog()

d.sound()

# Child class defines a method with the same name as the parent.
# The child version replaces (overrides) the parent version for that object.
```

```python
# ✅ 14. Use super()
class Animal:

    def sound(self):
        print("Animal Sound")

class Dog(Animal):

    def sound(self):
        super().sound()
        print("Dog Braks")

d  = Dog()
d.sound()

# super().sound() calls the parent class method.
# Then the child adds its own behavior.
```

```python
# ✅ 15. Parent Constructor
class Animal:

    def __init__(self):
        print("Animal Constructor")


class Dog(Animal):
    pass


d = Dog()
# Since Dog doesn't define its own constructor, it automatically uses the parent's constructor.
```

```python
# ✅ 16. Child Constructor with super()
class Animal:

    def __init__(self):
        print("animal constructor")

class Dog(Animal):

    def __init__(self):
        super().__init__()
        print("Dog Constructor")

d = Dog()
```

```python
# ✅ 17. Multilevel Inheritance
class A:

    def show(self):
        print("Class A")


class B(A):
    pass


class C(B):
    pass


obj = C()

obj.show()
```

```python
# Inheritance chain:

# A
# │
# B
# │
# C

# C inherits from B, and B inherits from A, so C can use show().
```

```python
# ✅ 18. Multiple Inheritance
class Father:

    def bike(self):
        print("Bike")


class Mother:

    def car(self):
        print("Car")


class Child(Father, Mother):
    pass


c = Child()

c.bike()
c.car()
# Child inherits from both Father and Mother.
```

```python
# ✅ 19. Polymorphism

class Dog:

    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

animals = [Dog(),Cat()]

for animal in animals:
    animal.sound()

# Both classes implement a sound() method. The same method call (animal.sound()) behaves differently depending on the object's type.
```

```python
# ✅ 20. Check Inheritance
class Animal:
    pass

class Dog(Animal):
    pass

print(isinstance(Dog(),Animal))
print(issubclass(Dog,Animal))

# isinstance(obj, Class) checks whether an object is an instance of a class (or its subclasses).
# issubclass(SubClass, ParentClass) checks whether one class inherits from another.
```

```python
# ✅ 21. Encapsulation (Private Variable)
class Student:

    def __init__(self):
        self.__marks = 90

    def show_marks(self):
        print(self.__marks)

s = Student()

s.show_marks()

# __marks is a private variable.
# It cannot be accessed directly outside the class.
```

```python
# ✅ 22. Access Private Variable (Wrong Way)
class Student:

    def __init__(self):
        self.__marks = 90

s = Student()

print(s.__marks)
# Private variables are name-mangled and cannot be accessed directly.
```

```python
# ✅ 23. Getter Method
class Student:

    def __init__(self):
        self.__marks = 90

    def get_marks(self):
        return self.__marks

s = Student()

print(s.get_marks())
```

```python
# ✅ 24. Setter Method
class Student:

    def __init__(self):
        self.__marks = 0

    def set_marks(self, marks):
        self.__marks = marks

    def get_marks(self):
        return self.__marks

s = Student()

s.set_marks(95)

print(s.get_marks())
```

```python
# ✅ 25. Static Method
class Math:

    @staticmethod
    def add(a, b):
        return a + b

print(Math.add(10, 20))
# A static method does not use self or cls.
```

```python
# ✅ 26. Class Method
class Student:

    school = "ABC School"

    @classmethod
    def show_school(cls):
        print(cls.school)

Student.show_school()
# A class method works with class variables using cls.
```

```python
# ✅ 27. Property Decorator
class Student:

    def __init__(self):
        self._marks = 90

    @property
    def marks(self):
        return self._marks

s = Student()

print(s.marks)
# @property lets you access a method like an attribute.
```

```python
# ✅ 28. Property Setter
class Student:

    def __init__(self):
        self._marks = 90

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        self._marks = value

s = Student()

s.marks = 100

print(s.marks)
```

```python
# ✅ 29. Abstract Class
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):

    def sound(self):
        print("Bark")

d = Dog()

d.sound()
# An abstract class defines methods that child classes must implement.
```

```python
# ✅ 30. Cannot Create Object of Abstract Class
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

a = Animal()
# You cannot instantiate a class that still contains abstract methods.
```

```python
# ✅ 31. Method Resolution Order (MRO)
class A:
    pass

class B(A):
    pass

class C(B):
    pass

print(C.mro())
# mro() shows the order Python follows when searching for methods.
```

```python
# ✅ 32. __str__() Method
class Student:

    def __str__(self):
        return "Student Object"

s = Student()

print(s)
# __str__() defines how an object is displayed when printed.
```

```python
# ✅ 33. __len__() Method
class Box:

    def __len__(self):
        return 5

b = Box()

print(len(b))
```

```python
# ✅ 34. Operator Overloading
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
# By defining __add__(), you control how the + operator behaves for your objects.
```
